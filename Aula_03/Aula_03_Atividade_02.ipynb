Atividade 2 — Investigação da Queda de Vendas

```python
%sql
-- Parte 1:Evolução mensal
-- Ano;
-- Mês;
-- Receita total;
-- Lucro total;
-- Quantidade vendida;

SELECT
    YEAR(Data) AS Ano,
    MONTH(Data) AS Mes,
    DATE_FORMAT(Data, 'yyyy-MM') AS Ano_Mes,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    SUM(Quantidade) AS Quantidade_Vendida
FROM workspace.default.vendas
GROUP BY YEAR(Data), MONTH(Data), DATE_FORMAT(Data, 'yyyy-MM')
ORDER BY Ano, Mes
```

```python
%sql
-- Parte 2: Identificação dos piores meses

SELECT
    YEAR(Data) AS Ano,
    MONTH(Data) AS Mes,
    DATE_FORMAT(Data, 'yyyy-MM') AS Ano_Mes,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    SUM(Quantidade) AS Quantidade_Vendida
FROM workspace.default.vendas
GROUP BY YEAR(Data), MONTH(Data), DATE_FORMAT(Data, 'yyyy-MM')
ORDER BY Receita_Total ASC
LIMIT 10
```

#### Qual foi o mês com menor receita?

### Dados:
- Receita Total: R$ 1.685.475,98
- Lucro Total: R$ 489.641,15
- Quantidade Vendida: 1.324 unidades
- Este mês apresentou a menor receita de todo o período analisado (19 meses de dados, de Janeiro/2025 a Julho/2026).

```python
%sql
-- Parte 3: Investigação por estado

SELECT
    Estado,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    SUM(Quantidade) AS Quantidade_Vendida,
    COUNT(*) AS Numero_Transacoes
FROM workspace.default.vendas
WHERE DATE_FORMAT(Data, 'yyyy-MM') = (
    SELECT DATE_FORMAT(Data, 'yyyy-MM')
    FROM workspace.default.vendas
    GROUP BY DATE_FORMAT(Data, 'yyyy-MM')
    ORDER BY SUM(Receita) ASC
    LIMIT 1
)
GROUP BY Estado
ORDER BY Receita_Total ASC
```

```python
%sql
-- Parte 4: Investigação por categoria

SELECT
    Estado,
    Categoria,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    SUM(Quantidade) AS Quantidade_Vendida,
    COUNT(*) AS Numero_Transacoes
FROM workspace.default.vendas
WHERE DATE_FORMAT(Data, 'yyyy-MM') = (
    SELECT DATE_FORMAT(Data, 'yyyy-MM')
    FROM workspace.default.vendas
    GROUP BY DATE_FORMAT(Data, 'yyyy-MM')
    ORDER BY SUM(Receita) ASC
    LIMIT 1
)
GROUP BY Estado, Categoria
ORDER BY Receita_Total ASC
LIMIT 20
```

```python
%sql
-- Parte 5: Investigação por vendedor

WITH pior_mes AS (
    SELECT DATE_FORMAT(Data, 'yyyy-MM') AS mes
    FROM workspace.default.vendas
    GROUP BY DATE_FORMAT(Data, 'yyyy-MM')
    ORDER BY SUM(Receita) ASC
    LIMIT 1
),
estado_problematico AS (
    SELECT Estado
    FROM workspace.default.vendas
    WHERE DATE_FORMAT(Data, 'yyyy-MM') = (SELECT mes FROM pior_mes)
    GROUP BY Estado
    ORDER BY SUM(Receita) ASC
    LIMIT 1
)
SELECT
    Vendedor,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    SUM(Quantidade) AS Quantidade_Vendida,
    COUNT(*) AS Numero_Transacoes,
    ROUND(AVG(Receita), 2) AS Ticket_Medio
FROM workspace.default.vendas
WHERE DATE_FORMAT(Data, 'yyyy-MM') = (SELECT mes FROM pior_mes)
  AND Estado = (SELECT Estado FROM estado_problematico)
GROUP BY Vendedor
ORDER BY Receita_Total ASC
```
