Atividade 1 — Diagnóstico de Desempenho Comercial

```python
%sql
-- Análise completa de desempenho por categoria
-- Receita total;
-- Custo total;
-- Lucro total;
-- Margem de lucro (%);
-- Quantidade total vendida;
-- Ticket médio das vendas;

SELECT
    Categoria,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Custo), 2) AS Custo_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS Margem_Lucro_Percentual,
    SUM(Quantidade) AS Quantidade_Total_Vendida,
    ROUND(AVG(Receita), 2) AS Ticket_Medio,
    CASE
        WHEN (SUM(Lucro) / SUM(Receita)) * 100 >= 30 THEN 'Alta Margem'
        WHEN (SUM(Lucro) / SUM(Receita)) * 100 >= 15 THEN 'Margem Média'
        ELSE 'Baixa Margem'
    END AS Classificacao_Margem
FROM workspace.default.vendas
GROUP BY Categoria
ORDER BY Margem_Lucro_Percentual DESC
```

```python
%sql
-- Visualização: Receita x Lucro por Categoria

SELECT
    Categoria,
    ROUND(SUM(Receita), 2) AS Receita_Total,
    ROUND(SUM(Lucro), 2) AS Lucro_Total,
    ROUND((SUM(Lucro) / SUM(Receita)) * 100, 2) AS Margem_Lucro_Percentual
FROM workspace.default.vendas
GROUP BY Categoria
ORDER BY Margem_Lucro_Percentual DESC
```

## Atividade 1

### Categoria com maior receita é necessariamente a mais rentável?

- Não necessariamente. Uma categoria pode ter alta receita mas baixa margem de lucro, resultando em menor rentabilidade comparada a categorias com menor receita mas margens superiores.

### Qual categoria deveria receber maior atenção da diretoria?

**Categoria que Merece Maior Atenção:**
   - A categoria que deve receber maior atenção é aquela com **baixa margem de lucro**, independentemente da receita. Essas categorias representam risco ao negócio, geram volume mas pouco retorno efetivo. É necessário investigar: custos elevados, preços baixos ou mix de produtos inadequado.
